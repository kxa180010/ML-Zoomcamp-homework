```python
asia_cars = df[df['origin'] == 'Asia']

columns = asia_cars[['vehicle_weight', 'model_year']]

first_seven = columns.head(7)
first_seven
```

```python
X = first_seven.to_numpy()
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
invX = np.linalg.inv(XTX)
invX
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w = invX.dot(X.T).dot(y)
w
```

```python
np.sum(w)
```
