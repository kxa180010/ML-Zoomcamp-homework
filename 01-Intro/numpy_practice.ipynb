Rndomly generate arrays

```python
np.random.seed(10)  #preduce the same number random 
np.random.rand(5,2)  #generate a random array with 5 rows and 2 columns sample from Uniform distribution
```

```python
np.random.seed(10)  #preduce the same number random 
np.random.randn(5,2)  #generate a random array with 5 rows and 2 columns sample from Standard Normal
```

```python
np.random.seed(10)  #preduce the same number random 
100*np.random.rand(5,2)  #generate a random array from 0 to 100 with 5 rows and 2 columns
```

```python
np.random.seed(10)  #preduce the same number random 
np.random.randint(low=0, high=100, size=(5,2))  #generate a random array of integers with 5 rows and 2 columns
```
